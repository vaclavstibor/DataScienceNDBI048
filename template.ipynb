```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import TruncatedSVD
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler
from sklearn.neighbors import KNeighborsRegressor
from sklearn.metrics import mean_squared_error, mean_absolute_error
from annoy import AnnoyIndex
import pandas as pd
import numpy as np
import joblib

# Load the dataset
# Ensure `df` is already preloaded with your data
df['combined_features'] = (
    df['primaryTitle'].fillna('') + ' ' +
    df['genres'].fillna('') + ' ' +
    df['directors'].fillna('') + ' ' +
    df['writers'].fillna('')
)

# TF-IDF Vectorization with fewer features
vectorizer = TfidfVectorizer(max_features=1000)
tfidf_matrix = vectorizer.fit_transform(df['combined_features'])

# Dimensionality Reduction
svd = TruncatedSVD(n_components=100, random_state=42)
reduced_matrix = svd.fit_transform(tfidf_matrix)

# Scale numeric features
numeric_features = ['runtimeMinutes', 'numVotes', 'averageRating']
df[numeric_features] = df[numeric_features].fillna(0)
scaler = MinMaxScaler()
scaled_numeric_features = scaler.fit_transform(df[numeric_features])

# Apply weight to numeric features
numeric_weight = 2.0  # Assign a higher weight to numeric features
scaled_numeric_features_weighted = scaled_numeric_features * numeric_weight

# Apply weight to TF-IDF features (genres)
#genre_weight = 0.5  # Lighter weight for genre features
#genre_indices = [vectorizer.get_feature_names_out().tolist().index(genre) for genre in ['action', 'comedy', 'drama']]  # Example genres
#for idx in genre_indices:
#    reduced_matrix[:, idx] *= genre_weight  # Apply the weight to the genre columns in the reduced matrix

# Combine all features
combined_matrix = np.hstack([reduced_matrix, scaled_numeric_features_weighted])

# Split data
X_train, X_test, y_train, y_test = train_test_split(
    combined_matrix, df['averageRating'], test_size=0.2, random_state=42
)

# Train KNN regressor
knn = KNeighborsRegressor(n_neighbors=5)
knn.fit(X_train, y_train)

# Save the model
#joblib.dump(knn, '/Users/stiborv/Documents/NMgr/ZS2425/NDBI048/ipynbs/knn_model_weighted.joblib')


# Predictions and evaluation
y_pred = knn.predict(X_test)
mse = mean_squared_error(y_test, y_pred)
mae = mean_absolute_error(y_test, y_pred)

print(f"Validation Mean Squared Error: {mse:.2f}")
print(f"Validation Mean Absolute Error: {mae:.2f}")

#Validation Mean Squared Error: 0.04
#Validation Mean Absolute Error: 0.10
```

```python
from annoy import AnnoyIndex

# Annoy Index for Fast Similarity Search
dimension = combined_matrix.shape[1]
annoy_index = AnnoyIndex(dimension, metric='angular')
for i, vector in enumerate(combined_matrix):
    annoy_index.add_item(i, vector)
annoy_index.build(10)

# Define the dimension of the reduced matrix
dimension = reduced_matrix.shape[1]

# Find Similar Movies Function
def find_similar_movies_annoy(movie_id, k=5):
    # Check if movie ID exists
    if movie_id not in df['tconst'].values:
        print(f"Movie ID {movie_id} not found in the dataset.")
        return None

    # Get the target movie's startYear
    target_start_year = df[df['tconst'] == movie_id]['startYear'].values[0]

    # Filter movies by startYear condition
    filtered_df = df[df['startYear'] <= target_start_year].reset_index(drop=True)

    # Create a filtered Annoy index
    filtered_index = AnnoyIndex(dimension, metric='angular')
    for i, vector in enumerate(reduced_matrix[filtered_df.index]):
        filtered_index.add_item(i, vector)

    filtered_index.build(10)

    # Get the movie's index in the filtered dataset
    movie_index = filtered_df[filtered_df['tconst'] == movie_id].index[0]

    # Find similar movies
    similar_indices = filtered_index.get_nns_by_item(movie_index, k + 1)
    similar_indices = [i for i in similar_indices if filtered_df.iloc[i]['tconst'] != movie_id][:k]

    if not similar_indices:
        print(f"No similar movies found for {movie_id}.")
        return None

    similar_movies = filtered_df.iloc[similar_indices]
    return similar_movies

# Example Usage
#movie_id = 'tt0944947'  # Replace with your target movie ID
#movie_id = 'tt3720452'   # Amazing Spiderman 2
movie_id = 'tt0435761'   # Toy Story 3
similar_movies = find_similar_movies_annoy(movie_id, k=5)

if similar_movies is not None:
    #print(similar_movies)
    # Print it as normal dataframe
    display(similar_movies)

# Display the movie details
display(df[df['tconst'] == movie_id])
```

## Weighted Directors and Writers

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import TruncatedSVD
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler
from sklearn.neighbors import KNeighborsRegressor
from sklearn.metrics import mean_squared_error, mean_absolute_error
from collections import Counter
import pandas as pd
import numpy as np
import joblib

# Load the dataset
# Ensure `df` is already preloaded with your data
df['combined_features'] = (
    df['primaryTitle'].fillna('') + ' ' +
    df['genres'].fillna('') + ' ' +
    df['directors'].fillna('') + ' ' +
    df['writers'].fillna('')
)

# TF-IDF Vectorization with fewer features
vectorizer = TfidfVectorizer(max_features=1000)
tfidf_matrix = vectorizer.fit_transform(df['combined_features'])

# Dimensionality Reduction
svd = TruncatedSVD(n_components=100, random_state=42)
reduced_matrix = svd.fit_transform(tfidf_matrix)

# Scale numeric features
numeric_features = ['runtimeMinutes', 'numVotes', 'averageRating']
df[numeric_features] = df[numeric_features].fillna(0)
scaler = MinMaxScaler()
scaled_numeric_features = scaler.fit_transform(df[numeric_features])

def get_weighted_directors_writers(df, column_name, rating_column='averageRating'):
    all_values = df[column_name].fillna('').str.split(',').explode().str.strip()
    rating_map = df.groupby(column_name)[rating_column].mean()  # Get average ratings for each director/ writer
    weight_map = rating_map.to_dict()  # Map of director/writer to average rating

    weighted_column = df[column_name].fillna('').apply(
        lambda x: np.mean([weight_map.get(val.strip(), 0) for val in x.split(',')]) if x else 0
    )

    return weighted_column


# Calculate weights for directors and writers
df['directors_weight'] = get_weighted_directors_writers(df, 'directors')
df['writers_weight'] = get_weighted_directors_writers(df, 'writers')

# Combine weights into a single feature
df['directors_writers_weighted'] = df['directors_weight'] + df['writers_weight']

# Now incorporate these weights into the final feature set
scaled_numeric_features_weighted = scaled_numeric_features * df['directors_writers_weighted'].values[:, np.newaxis]

# Combine all features
combined_matrix = np.hstack([reduced_matrix, scaled_numeric_features_weighted])

# Split data
X_train, X_test, y_train, y_test = train_test_split(
    combined_matrix, df['averageRating'], test_size=0.2, random_state=42
)

# Train KNN regressor
knn = KNeighborsRegressor(n_neighbors=5)
knn.fit(X_train, y_train)

# Predictions and evaluation
y_pred = knn.predict(X_test)
mse = mean_squared_error(y_test, y_pred)
mae = mean_absolute_error(y_test, y_pred)

print(f"Validation Mean Squared Error: {mse:.2f}")
print(f"Validation Mean Absolute Error: {mae:.2f}")
```

```python
df = df.assign(genres=df['genres'].str.split(',')).explode('genres')

# Filter data for the period from 1990 to 2024
df_filtered = df[(df['startYear'] >= 1990) & (df['startYear'] <= 2024)]

# Group by year and genre, then calculate the average rating
df_grouped = df_filtered.groupby(['startYear', 'genres']).mean().reset_index()

# Create the line plot
fig_line = px.line(
    df_grouped,
    x='startYear',
    y='averageRating',
    color='genres',
    title='Average Movie Ratings Over Time by Genre',
    labels={'startYear': 'Year', 'averageRating': 'Average Rating', 'genres': 'Genre'}
)

# Add interactivity and styling
fig_line.update_layout(
    xaxis_title='Year',
    yaxis_title='Average Rating',
    legend_title='Genre',
    autosize=True,
    margin=dict(l=50, r=50, t=50, b=50)
)

# Display the interactive line plot
fig_line.show()
```
